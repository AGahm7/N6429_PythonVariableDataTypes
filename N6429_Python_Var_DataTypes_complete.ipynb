## Review Python Strings

Review the pages in the __[Python Strings](https://www.w3schools.com/python/python_strings.asp)__ section.

#### Code Challenge

**Instructions**

Inside the editor, complete the following steps:

Print the characters from index 2 to 5 (slicing)

Print txt converted to upper case

Use an f-string to print "I love Python" using the name variable

````python
# Create the variable
txt = 'Hello, World!'
# Print characters from index 2 to 5
print(txt[2:5])
# Print in upper case
print(txt.upper())
# Create the name variable
name = 'Python'
# Print using an f-string
print(f"I love {name}")
````
